## 3. Condicionales

- Python usa operadores booleanos para evaluar condiciones.
- Operadores de comparación: `==`, `>=`, `<=`, `>`, `<`, `!=`.
- Operadores lógicos: `not`, `and`, `or`.
- Si la condición del `if` es verdadera, se ejecuta ese bloque. Si no, Python evalúa el siguiente `elif` (si existe); si ninguna condición se cumple, ejecuta el bloque `else`.

Sintaxis:
```python
if expresion:
    instruccion_1
elif otra_expresion:
    instruccion_2
else:
    instruccion_3
```

## 4. Loops (bucles)

Un bucle `for` repite un bloque de código para cada elemento de una secuencia.

````python
mi_lista = ["Hola", "mundo", "!"]

for cosa in mi_lista:
    print(cosa)
````

````python
for numero in range(0, 4):
    print(numero)
````

`zip()` permite iterar en paralelo sobre dos o más secuencias. `enumerate()` da a la vez el índice y el valor.

````python
frutas = ["manzana", "plátano", "mango"]
verduras = ["zanahoria", "papa", "tomate"]

for f, v in zip(frutas, verduras):
    print("Fruta:", f, "| Verdura:", v)

for i, valor in enumerate(frutas):
    print(i, valor)
````

Para resolver el siguiente ejercicio vas a necesitar **acumular** un resultado a lo largo del bucle: una variable que empieza en un valor inicial, y se va actualizando en cada vuelta. Por ejemplo, para sumar los primeros 5 números:

````python
suma = 0
for numero in range(1, 6):
    suma = suma + numero  # forma larga
print(suma)
````

`suma = suma + numero` se puede escribir de forma abreviada como `suma += numero`: toma el valor actual de `suma`, le suma `numero`, y guarda el resultado de vuelta en `suma`. Existen las mismas abreviaturas para el resto de operadores: `-=`, `*=`, `/=`.

````python
suma = 0
for numero in range(1, 6):
    suma += numero  # forma abreviada, mismo resultado
print(suma)
````
